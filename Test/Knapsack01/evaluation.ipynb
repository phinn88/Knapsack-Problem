```python
# Function to read and print the content of a 0-1 Knapsack test file
def print_knapsack_test(file_path):
    try:
        with open(file_path, 'r') as file:
            lines = file.readlines()
            # Dòng đầu tiên: số lượng item và sức chứa tối đa
            n, maxweight = map(int, lines[0].split())
            print(f"Number of items: {n}, Max weight: {maxweight}")
            
            # Các dòng tiếp theo: giá trị và trọng lượng của từng item
            print("Items (value, weight):")
            for line in lines[1:]:
                try:
                    value, weight = map(int, line.split())
                    print(f"  Value: {value}, Weight: {weight}")
                except ValueError:
                    # Nếu không đọc được giá trị và trọng lượng, có thể là dòng "choose"
                    print(f"  Additional info: {line.strip()}")
    except Exception as e:
        print(f"Error reading file {file_path}: {e}")

# Main function
if __name__ == "__main__":
    file_path = r"D:\Knapsack-Problem\Test\Data\instances_01_KP\large_scale\knapPI_1_100_1000_1"
    print(f"Reading file: {file_path}")
    print_knapsack_test(file_path)
```

```python
import time

# Greedy algorithm for 0-1 Knapsack Problem
def knapsack_greedy_by_value(values, weights, capacity):
    items = sorted(zip(values, weights), key=lambda x: x[0], reverse=True)  # Sắp xếp theo giá trị giảm dần
    total_value = 0
    for value, weight in items:
        if capacity >= weight:
            capacity -= weight
            total_value += value
    return total_value

def knapsack_greedy_by_weight(values, weights, capacity):
    items = sorted(zip(values, weights), key=lambda x: x[1])  # Sắp xếp theo trọng lượng tăng dần
    total_value = 0
    for value, weight in items:
        if capacity >= weight:
            capacity -= weight
            total_value += value
    return total_value

def knapsack_greedy_by_ratio(values, weights, capacity):
    items = sorted(zip(values, weights), key=lambda x: x[0] / x[1], reverse=True)  # Sắp xếp theo tỷ lệ giá trị/trọng lượng giảm dần
    total_value = 0
    for value, weight in items:
        if capacity >= weight:
            capacity -= weight
            total_value += value
    return total_value

def knapsack_greedy_combined(values, weights, capacity):
    # Sắp xếp theo giá trị giảm dần
    items_by_value = sorted(zip(values, weights), key=lambda x: x[0], reverse=True)
    total_value_by_value = 0
    remaining_capacity = capacity
    for value, weight in items_by_value:
        if remaining_capacity >= weight:
            remaining_capacity -= weight
            total_value_by_value += value

    # Sắp xếp theo trọng lượng tăng dần
    items_by_weight = sorted(zip(values, weights), key=lambda x: x[1])
    total_value_by_weight = 0
    remaining_capacity = capacity
    for value, weight in items_by_weight:
        if remaining_capacity >= weight:
            remaining_capacity -= weight
            total_value_by_weight += value

    # Sắp xếp theo tỷ lệ giá trị/trọng lượng giảm dần
    items_by_ratio = sorted(zip(values, weights), key=lambda x: x[0] / x[1], reverse=True)
    total_value_by_ratio = 0
    remaining_capacity = capacity
    for value, weight in items_by_ratio:
        if remaining_capacity >= weight:
            remaining_capacity -= weight
            total_value_by_ratio += value

    # Chọn kết quả lớn nhất giữa hai cách sắp xếp
    return max(total_value_by_weight, total_value_by_ratio, total_value_by_value)

# Dynamic Programming algorithm for 0-1 Knapsack Problem
def knapsack_dp(values, weights, capacity):
    n = len(values)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]
    for i in range(1, n + 1):
        for w in range(capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(dp[i - 1][w], dp[i - 1][w - weights[i - 1]] + values[i - 1])
            else:
                dp[i][w] = dp[i - 1][w]
    return dp[n][capacity]

# Backtracking algorithm with memoization for 0-1 Knapsack Problem
import random

# Genetic Algorithm for 0-1 Knapsack Problem
def knapsack_genetic(values, weights, capacity, population_size=100, generations=100, mutation_rate=0.01):
    n = len(values)

    # Hàm khởi tạo quần thể ban đầu
    def initialize_population():
        population = []
        for _ in range(population_size):
            while True:
                chromosome = [random.randint(0, 1) for _ in range(n)]
                if is_valid(chromosome):  # Đảm bảo cá thể hợp lệ
                    population.append(chromosome)
                    break
        return population

    # Hàm kiểm tra tính hợp lệ của một cá thể
    def is_valid(chromosome):
        total_weight = sum(w * c for w, c in zip(weights, chromosome))
        return total_weight <= capacity

    # Hàm tính độ thích nghi (fitness)
    def fitness(chromosome):
        total_value = sum(v * c for v, c in zip(values, chromosome))
        total_weight = sum(w * c for w, c in zip(weights, chromosome))
        if total_weight > capacity:
            return 0  # Nếu vượt quá sức chứa, độ thích nghi bằng 0
        return total_value

    # Hàm chọn lọc (selection) - Chọn cá thể tốt nhất
    def selection(population):
        population.sort(key=fitness, reverse=True)
        return population[:population_size // 2]  # Chọn nửa trên của quần thể

    # Hàm lai ghép (crossover)
    def crossover(parent1, parent2):
        point = random.randint(1, n - 1)  # Chọn điểm lai ghép
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
        return child1, child2

    # Hàm đột biến (mutation)
    def mutate(chromosome):
        for i in range(n):
            if random.random() < mutation_rate:
                chromosome[i] = 1 - chromosome[i]  # Đảo bit (0 -> 1, 1 -> 0)

    # Khởi tạo quần thể
    population = initialize_population()

    # Vòng lặp qua các thế hệ
    for generation in range(generations):
        # Đánh giá và chọn lọc
        population = selection(population)

        # Tạo thế hệ mới bằng lai ghép
        new_population = []
        while len(new_population) < population_size:
            parent1, parent2 = random.sample(population, 2)  # Chọn ngẫu nhiên 2 cá thể
            child1, child2 = crossover(parent1, parent2)
            mutate(child1)
            mutate(child2)
            # Đảm bảo cá thể hợp lệ trước khi thêm vào quần thể
            if is_valid(child1):
                new_population.append(child1)
            if is_valid(child2):
                new_population.append(child2)

        # Nếu quần thể mới chưa đủ, bổ sung thêm cá thể ngẫu nhiên hợp lệ
        while len(new_population) < population_size:
            chromosome = [random.randint(0, 1) for _ in range(n)]
            if is_valid(chromosome):
                new_population.append(chromosome)

        population = new_population

    # Trả về cá thể tốt nhất
    best_chromosome = max(population, key=fitness)
    best_value = fitness(best_chromosome)
    return best_value

# Evaluation function
def evaluate_knapsack_algorithms(values, weights, capacity):
    algorithms = {
        "Greedy_by_value": knapsack_greedy_by_value,
        "Greedy_by_weight": knapsack_greedy_by_weight,
        "Greedy_by_ratio": knapsack_greedy_by_ratio,
        "Greedy_combined": knapsack_greedy_combined,
        "Dynamic Programming": knapsack_dp,
        "Genetic":knapsack_genetic,
    }
    results = {}
    for name, algorithm in algorithms.items():
        start_time = time.time()
        result = algorithm(values, weights, capacity)
        elapsed_time = time.time() - start_time
        results[name] = {"Result": result, "Time (s)": elapsed_time}
    return results

# Example usage
if __name__ == "__main__":
    # Example data
    values = [60, 100, 120]
    weights = [10, 20, 30]
    capacity = 50

    # Evaluate algorithms
    results = evaluate_knapsack_algorithms(values, weights, capacity)
    for algo, metrics in results.items():
        print(f"{algo}: Result = {metrics['Result']}, Time = {metrics['Time (s)']:.6f} seconds")
```

```python
def parse_knapsack_file(file_path):
    try:
        with open(file_path, 'r') as file:
            lines = file.readlines()
            
            # Dòng đầu tiên: số lượng item và sức chứa tối đa
            n_maxweight = list(map(int, lines[0].split()))
            
            # Các dòng tiếp theo: giá trị và trọng lượng của từng item
            values = []
            weights = []
            
            # Duyệt qua các dòng, trừ dòng cuối cùng
            for line in lines[1:-1]:
                value, weight = map(int, line.split())
                values.append(value)
                weights.append(weight)
            
            # Dòng cuối cùng chứa mảng nhị phân
            how_to_choose = list(map(int, lines[-1].strip().split()[1:]))
            
            return n_maxweight, values, weights, how_to_choose
    except Exception as e:
        print(f"Error reading file {file_path}: {e}")
        return None, None, None, None    
```

```python
def read_optimum(file_path):
    try:
        with open(file_path, 'r') as file:
            max_value = int(file.readline().strip())  # Đọc dòng đầu tiên và chuyển thành số nguyên
            return max_value
    except Exception as e:
        print(f"Error reading optimum file {file_path}: {e}")
        return None
```

```python
def parse_knapsack_file_low(file_path):
    try:
        with open(file_path, 'r') as file:
            lines = file.readlines()
            
            # Dòng đầu tiên: số lượng item và sức chứa tối đa
            n_maxweight = list(map(int, lines[0].split()))
            
            # Các dòng tiếp theo: giá trị và trọng lượng của từng item
            values = []
            weights = []
            
            # Duyệt qua các dòng, trừ dòng cuối cùng
            for line in lines[1:]:
                value, weight = map(int, line.split())
                values.append(value)
                weights.append(weight)
        
            return n_maxweight, values, weights
    except Exception as e:
        print(f"Error reading file {file_path}: {e}")
        return None, None, None   
```

```python
import os

# Function to list all files in a directory
def list_files_in_directory(directory_path):
    try:
        # Lấy danh sách tất cả các file trong thư mục
        files = [file for file in os.listdir(directory_path) if os.path.isfile(os.path.join(directory_path, file))]
        return files
    except Exception as e:
        print(f"Error accessing directory {directory_path}: {e}")
        return []
```
